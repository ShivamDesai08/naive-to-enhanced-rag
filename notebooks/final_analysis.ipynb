## 2. Performance Comparison


```python
# Create performance comparison visualization
if 'naive_results' in locals() and 'enhanced_results' in locals():
    metrics = ['f1', 'exact_match', 'avg_context_length', 'avg_response_length']
    
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    axes = axes.ravel()
    
    for i, metric in enumerate(metrics):
        if metric in naive_results.columns and metric in enhanced_results.columns:
            data = [naive_results[metric].mean(), enhanced_results[metric].mean()]
            axes[i].bar(['Naive RAG', 'Enhanced RAG'], data, color=['skyblue', 'lightcoral'])
            axes[i].set_title(f'{metric.replace("_", " ").title()}')
            axes[i].set_ylabel('Score')
    
    plt.tight_layout()
    plt.show()
else:
    print("No results data available for visualization")
    
    # Example visualization data
    print("\nExample Performance Comparison:")
    print("Metric                | Naive RAG | Enhanced RAG | Improvement")
    print("---------------------|-----------|--------------|-------------")
    print("F1 Score             | 0.7692    | 0.8200       | +6.6%")
    print("Exact Match          | 0.0000    | 0.0500       | +5.0%")
    print("Avg Context Length   | 501.7     | 485.2        | -3.3%")
    print("Avg Response Length  | 1.67      | 2.15         | +28.7%")
```
